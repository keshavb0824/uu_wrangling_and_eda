# Top 20 schools

# Flow chart analysis

```python
import plotly.graph_objects as go


giftor = 'Giftor Name'
#giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 20

flows = (
    duke_df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```

All these schools are located on the east coast and are all academically well known institutions. Cornell and Harvard are Ivy league schools while Carnegie Mellon and MIT are extremely well known institutions. UVA should look into how all those schools are receiving these donations and see if it could access them as well.
